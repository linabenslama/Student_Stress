### Filtrage simple du jeu de données
Les étudiants qui révisent beaucoup (6 heures ou plus)

```python
df[df['study_hours'] >= 6]
```

Les étudiants qui sont très stressés

```python
df[df['stress_level'] == 'High']
```

### Filtrage avec conditions multiples

Les étudiants qui ont des mauvaises notes ET le stress élevé

```python
df[(df['exam_score']<50) & (df['stress_level'] == 'High')]
```

Top étudiants (peu de stress + bonnes notes)

```python
df[(df['stress_level'] == 'Low') & (df['exam_score'] > 75)]
```

Les étudiants qui utilisent beaucoup les réseaux sociaux OU manquent de sommeil

```python
df[(df['social_media_hours'] > 4) | (df['sleep_hours'] < 6)]
```

### Filtrage par catégorie

Les étudiants qui ont 7 ou 8 heures de sommeil

```python
df [df['sleep_hours'].isin([7, 8])]
```

### Tri des données

Tri décroissant par notes (Top 5 étudiants)

```python
df.sort_values(by='exam_score', ascending=False).head()
```

Tri croissant par heures de réseaux sociaux + niveau de stress

```python
df.sort_values(by=['social_media_hours','stress_level'], ascending=[True, True])
```

Tri par niveau de stress et afficher les heures utilisées en réseaux sociaux, les notes et niveau de stress

```python
df[['social_media_hours', 'exam_score', 'stress_level']].sort_values(by=['stress_level'])
```

### Distribution des niveaux de stress

```python
df['stress_level'].value_counts() #répartition du stress
```

### Distribution des niveaux de stress en pourcentage

```python
df['stress_level'].value_counts(normalize=True)
```

### Moyenne des notes des étudiants par nombre d'heures d'études

```python
df.groupby('study_hours')['exam_score'].mean() 
```

### Moyenne des notes des étudiants par le niveau du stress

```python
df.groupby('stress_level')['exam_score'].mean()
```

### Moyenne des notes des étudiants par nombre d'heures de sommeil et de réseaux sociaux

```python
df.groupby(['sleep_hours', 'social_media_hours'])['exam_score'].mean()
```

### Moyenne des notes des étudiants par heures de sommeil, heures de réseaux sociaux et niveau de stress

```python
df.groupby(['study_hours', 'sleep_hours', 'social_media_hours', 'stress_level'])['exam_score'].mean()
```
